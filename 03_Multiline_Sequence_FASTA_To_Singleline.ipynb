```python
def flatten_fasta(input_file, output_file):
    with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
        sequence_parts = []
        header = None
        
        for line in infile:
            line = line.strip()
            if line.startswith('>'):
                if header is not None:
                    outfile.write(header + '\n')
                    outfile.write(''.join(sequence_parts) + '\n')
                header = line
                sequence_parts = []
            else:
                sequence_parts.append(line)
        
        if header is not None:
            outfile.write(header + '\n')
            outfile.write(''.join(sequence_parts) + '\n')

# Example usage
flatten_fasta("multiline_input.fasta", "flattened_output.fasta")
```
