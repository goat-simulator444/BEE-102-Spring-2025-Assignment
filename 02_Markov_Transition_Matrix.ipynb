using a demo sequence to generate the markov transition matrix

```python
sequence = "ATGCGATGACCTGACTTTAAGGCTAGGCTAAGCCAATTGGCCATGCCATTGGC"

# Ensure the sequence is uppercase and remove invalid characters (optional)
valid_nucleotides = {'A', 'T', 'G', 'C'}
sequence = ''.join([n for n in sequence.upper() if n in valid_nucleotides])

# Initialize the transition count matrix
nucleotides = ['A', 'T', 'G', 'C']
transition_counts = {n1: {n2: 0 for n2 in nucleotides} for n1 in nucleotides}

# Count transitions
for i in range(len(sequence) - 1):
    current = sequence[i]
    next_ = sequence[i + 1]
    transition_counts[current][next_] += 1

# Convert counts to probabilities (transition matrix)
transition_matrix = {}
for current in nucleotides:
    total = sum(transition_counts[current].values())
    transition_matrix[current] = {}
    for next_ in nucleotides:
        if total > 0:
            transition_matrix[current][next_] = transition_counts[current][next_] / total
        else:
            transition_matrix[current][next_] = 0.0

# Display the transition matrix
import pandas as pd
df = pd.DataFrame(transition_matrix)
print("Transition Matrix (rows = current nucleotide, columns = next nucleotide):")
display(df.T)  # Transposed for readability
```
